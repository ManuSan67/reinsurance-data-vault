```python
dbutils.widgets.text("catalog", "reinsurance")
catalog = dbutils.widgets.get("catalog")

schemas = {
    "landing": "Raw bordereaux files as sent by each cedent",
    "staging": "Typed source data with hash keys",
    "raw_vault": "Hubs, links and satellites",
    "business_vault": "PIT tables and business rules",
    "marts": "Star schema for reporting",
    "legacy": "Legacy ETL to be modernized",
}

spark.sql(f"CREATE CATALOG IF NOT EXISTS {catalog}")
for name, comment in schemas.items():
    spark.sql(f"CREATE SCHEMA IF NOT EXISTS {catalog}.{name} COMMENT '{comment}'")
spark.sql(f"CREATE VOLUME IF NOT EXISTS {catalog}.landing.bordereaux")

display(spark.sql(f"SHOW SCHEMAS IN {catalog}"))
```
